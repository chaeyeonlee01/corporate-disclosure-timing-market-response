````python
import FinanceDataReader as fdr
import pandas as pd

# Set the collection period (same as the stock price data)
START = "2018-07-01"
END   = "2026-04-30"

# KOSPI Index
kospi = fdr.DataReader("KS11", START, END)
kospi = kospi.reset_index()
kospi.columns = [c.lower() for c in kospi.columns]
kospi["index_type"] = "KOSPI"

# KOSDAQ Index
kosdaq = fdr.DataReader("KQ11", START, END)
kosdaq = kosdaq.reset_index()
kosdaq.columns = [c.lower() for c in kosdaq.columns]
kosdaq["index_type"] = "KOSDAQ"

# Combine the two market index datasets
index_df = pd.concat([kospi, kosdaq], ignore_index=True)

print(index_df.shape)
print(index_df.head())
````

````python
index_df.to_csv(
    "market_index_2018_2026.csv",
    index=False,
    encoding="utf-8-sig"
)
````

## 3. KOSPI and KOSDAQ Market Index Data Collection

### 3-1. Purpose of Data Collection

To calculate Abnormal Returns (AR) in the Event Study, information on overall market returns is required in addition to individual stock returns.

This study estimated expected returns using the Market Model and therefore collected KOSPI and KOSDAQ market index data.

The market index data were used for the following purposes:

```text
1. Calculate market returns
2. Estimate the Market Model regression
3. Calculate Abnormal Returns (AR)
4. Calculate Cumulative Abnormal Returns (CAR)
```

---

### 3-2. Collection Period

The market index data collection period was set to the same period as the individual stock price data:

```text
2018-07-01 ~ 2026-04-30
```

This period was selected to ensure sufficient data for the Event Study estimation window (t = -120 to -21).

---

### 3-3. Data Collection Method

Market index data were collected using the Python `FinanceDataReader` package.

The following market indices were collected:

| Market | Code |
|---|---|
| KOSPI | KS11 |
| KOSDAQ | KQ11 |

Daily market index data were downloaded for each index, and the column names were converted to lowercase for consistency and ease of analysis.

An `index_type` variable was also added to distinguish between KOSPI and KOSDAQ observations.

```python
kospi["index_type"] = "KOSPI"
kosdaq["index_type"] = "KOSDAQ"
```

The two market index datasets were then combined into a single DataFrame.

---

### 3-4. Collected Variables

The final market index dataset contained the following variables:

```text
date
open
high
low
close
volume
change
index_type
```

The `change` variable was used to measure daily market returns.

---

### 3-5. Data Storage

The collected market index data were saved as:

```text
market_index_2018_2026.csv
```

---

### 3-6. Role in the Analysis

The market index data served as the benchmark market return data for the Event Study.

Specifically, the data were used in the following procedures:

```text
1. Merge individual stock returns with market returns
2. Estimate the Market Model over the estimation window
3. Calculate Expected Returns
4. Calculate Abnormal Returns (AR)
5. Calculate Cumulative Abnormal Returns (CAR)
```

Different benchmark indices were assigned according to each firm's market classification (KOSPI/KOSDAQ).

For example, the KOSPI index was used as the benchmark market index for KOSPI-listed firms, while the KOSDAQ index was used for KOSDAQ-listed firms.

This approach was designed to estimate the effect of the disclosure itself on stock returns while controlling for overall market movements.
